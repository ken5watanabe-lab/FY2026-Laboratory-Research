# 1. TMM CommCohort health checkup 67K

## 1-1. Demographics

```python
#Import dataframe
fileDir = '../TMM/2024-0068/20251014_TMM_CommCohort_HealthCheckup_67K/cohort_data/utf8/'
fileName = 'demographics.csv'
tempDF = pd.read_csv(fileDir+fileName)

display(tempDF)
display(tempDF.describe(include='all'))
```

```python
#Clean
tempL = ['ParticipantID', 'BirthYear', 'BirthMonth', 'Sex', 'BaseCAge']
tempDF.columns = tempL
tempD = {1:'Male', 2:'Female'}
tempDF['Sex'] = tempDF['Sex'].map(tempD)

display(tempDF)
display(tempDF.describe(include='all'))

demographicsDF = tempDF
```

```python
tempDF = demographicsDF
variable = 'BaseCAge'
xaxis_label = 'Baseline age [years]'
tempD = {'All':'k', 'Female':'tab:red', 'Male':'tab:blue'}

#Check distribution
sns.set(style='ticks',  context='talk')
plt.figure(figsize=(4, 3))
##Group
for group in tempD.keys():
    if group=='All':
        tempS = tempDF[variable].dropna()
    else:
        tempS = tempDF[variable].loc[tempDF['Sex']==group].dropna()
    print(group+': skewness =', stats.skew(tempS))
    var_label = group+' ('+r'$n$'+' = '+f'{len(tempS):,}'+')'
    sns.distplot(tempS, label=var_label, color=tempD[group])
    plt.axvline(x=tempS.mean(), **{'linestyle':'--', 'color':tempD[group]})
sns.despine()
plt.ylabel('Density')
plt.xlabel(xaxis_label)
plt.legend(bbox_to_anchor=(1, 0.5), loc='center left', borderaxespad=1, labelspacing=0.25)
plt.show()
```

```python
tempDF = demographicsDF
variable = 'BirthYear'
xaxis_label = 'Birth year'
tempD = {'All':'k', 'Female':'tab:red', 'Male':'tab:blue'}

#Check distribution
sns.set(style='ticks',  context='talk')
plt.figure(figsize=(4, 3))
##Group
for group in tempD.keys():
    if group=='All':
        tempS = tempDF[variable].dropna()
    else:
        tempS = tempDF[variable].loc[tempDF['Sex']==group].dropna()
    print(group+': skewness =', stats.skew(tempS))
    var_label = group+' ('+r'$n$'+' = '+f'{len(tempS):,}'+')'
    sns.distplot(tempS, label=var_label, color=tempD[group])
    plt.axvline(x=tempS.mean(), **{'linestyle':'--', 'color':tempD[group]})
sns.despine()
plt.ylabel('Density')
plt.xlabel(xaxis_label)
plt.legend(bbox_to_anchor=(1, 0.5), loc='center left', borderaxespad=1, labelspacing=0.25)
plt.show()
```
