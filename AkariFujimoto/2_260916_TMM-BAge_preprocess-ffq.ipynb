```python
tempDF = ffqDF.copy()

#Clean
tempDF = tempDF.set_index('ID')
tempDF.index.name = 'ParticipantID'
tempL = ['予約番号','イベントコード','コースコード']
tempDF = tempDF.drop(columns=tempL)


print('Data:')
display(tempDF)
display(tempDF.describe(include='all'))

ffqDF = tempDF#Update
```

### -2. replace

```python
#検査項目コードとサフィックスを統合
tempDF=codeDF.copy()

tempDF['coding']=(tempDF['検査項目コード'].astype(str)+'.'+tempDF['サフィックス'].astype(str).str.zfill(2))

display(tempDF)

codeDF=tempDF
```

```python
#数字を文字に
tempDF1=ffqDF.copy()
tempDF2=codeDF

print('Before:', len(tempDF1.columns))

#検査項目名の置き換え
#int->str
tempDF=tempDF2['coding']

#dictionary
tempD=dict(zip(tempDF,tempDF2['検査項目名']))

tempL=[]
for col in tempDF1.columns:
    code=str(col)
    tempL.append(tempD.get(code,col))


tempDF1.columns=tempL
display(tempDF1)

print('After:', len(tempL))
tempL=[col for col in tempDF1.columns
       if col not in tempDF2['検査項目名'].values]

print('unmatched:')
if not tempL:
    print('All code matched.')
else:
    print(tempL)
    
ffqDF=tempDF1
```

```python
# name rename
tempDF = ffqDF.copy()

#old columns
tempD1 = tempDF.columns.tolist()

# Column名の空白を正規化
def normalize_col(col):
    col = col.replace('\u3000', ' ')  # 全角スペース → 半角スペース
    col = re.sub(r'\s+', ' ', col)    # 連続する空白を1つに
    col = col.strip()                 # 前後の空白を削除
    return col

# 既存DFの列名を置換
tempDF.columns = [normalize_col(col) for col in tempDF.columns]

#new columns
tempD2 = [normalize_col(col) for col in tempD1]

# check rename
tempDF1 = pd.DataFrame({'Original': tempD1, 'Normalized': tempD2})

display(tempDF1[tempDF1['Original'] != tempDF1['Normalized']])

# 正規化後に重複する列名があるか確認
tempI = pd.Index(tempD2)[pd.Index(tempD2).duplicated()].unique()

print('Duplicated columns after normalization:')
print(tempI.tolist())

if not pd.Index(tempD2).has_duplicates:
    tempDF.columns = tempD2
    display(tempDF)

    #fileDir = './savedata/'
    #ipynbName = '260916_TMM-BAge_CommCohort67K-ver1_'
    #fileName = 'ffqDF-rename-date.tsv'
    #tempDF = tempDF
    #tempDF.to_csv(fileDir+ipynbName+fileName, sep='\t', index=True)

else:
    print('Column name collision detected. Rename manually.')

ffqDF = tempDF
```

### -2 disease

```python
#回答者自身が発症しているもののみ
tempDF = lifeDF.copy()
tempDF = tempDF.filter(regex ='本人')

display(tempDF)
display(tempDF.describe(include='all'))

diDF = tempDF
```

```python
print(diDF.columns.tolist())
```

```python
tempDF = pd.DataFrame({
    '発症歴あり人数': diDF.eq(1).sum(),
    '有効回答数': diDF.notna().sum(),})

tempDF['発症歴あり割合(%)'] = (tempDF['発症歴あり人数'] / tempDF['有効回答数'] * 100).round(2)

tempDF = tempDF.sort_values('発症歴あり人数', ascending=False)

display(tempDF)

# 病名ごとのサブプロット
tempI=diDF.eq(1).sum().sort_values(ascending=False).index
ncols = 6
nrows = math.ceil(len(tempI) / ncols)

fig, axes = plt.subplots(
    nrows, ncols,
    figsize=(ncols * 3, nrows * 2.5),
    sharey=True)

axes = axes.flatten()

for i, col in enumerate(tempI):
    ax = axes[i]

    # 0/1の人数を集計（欠損値は除外）
    counts = diDF[col].value_counts().reindex([0, 1], fill_value=0)

    ax.bar(['0', '1'], counts.values)
    ax.set_title(f'{col}\n(n={int(counts.sum())}, 1={int(counts[1])})',
                 fontsize=8)
    ax.tick_params(axis='both', labelsize=8)

    # 1の人数を棒の上に表示
    ax.text(1, counts[1], str(int(counts[1])),
            ha='center', va='bottom', fontsize=8)

# 余ったサブプロットを削除
for j in range(len(tempI), len(axes)):
    fig.delaxes(axes[j])

fig.suptitle('Disease history (0/1)', fontsize=16)
plt.tight_layout()
plt.show()
```
