```python
cohortD = {'SexPooled': 'tab:green','Female': 'tab:red','Male': 'tab:blue'}

modelD = {
    'MetAge': met_s,
    'AntAge': ant_s,
    'HemAge': hem_s,
    'ChemAge': chem_s,
    'LabAge': lab_s,
    'AlgAge': alg_s,
    'KidAge': kid_s,}

model_colors = {
    'MetAge': 'tab:blue',
    'AntAge': 'tab:orange',
    'HemAge': 'tab:green',
    'ChemAge': 'tab:red',
    'LabAge': 'tab:purple',
    'AlgAge': 'tab:brown',
    'KidAge': 'tab:pink'}

# DF作成

met = {'R2score_P':'PAge', 'R2score_B':'BAge'}

for metric, yvar_model in met.items():
    tempL = []
    print('========================= '+yvar_model+' ==========================')
    for model, summary in modelD.items():
        for cohort in cohortD.keys():

            tempDF = summary[cohort][[metric]].copy()
            tempDF['Model'] = model
            tempDF['Cohort'] = cohort

            tempL.append(tempDF)

    tempDF = pd.concat(tempL, axis=0)
    tempDF = tempDF.rename(columns={metric: 'R2'})

    display(tempDF.groupby(['Cohort', 'Model'])['R2'].describe())


# Plot
    sns.set(style='ticks', context='talk')
    japanize_matplotlib.japanize()

    fig, axes = plt.subplots(
    1, 3,
    figsize=(15, 5),
    sharey=True)

    for ax, cohort in zip(axes, cohortD.keys()):
        plotDF = tempDF.loc[tempDF['Cohort'] == cohort].copy()

    # Bar
        sns.barplot(
        data=plotDF,
        y='R2',
        x='Model',
        order=modelD.keys(),
        hue='Model',
        hue_order=modelD.keys(),
        palette=model_colors,
        dodge=False,
        ci=95,
        capsize=0.2,
        errwidth=1.2,
        alpha=0.7,
        errcolor='black',
        edgecolor='black',
        ax=ax)

    # Individual folds
        sns.stripplot(
        data=plotDF,
        y='R2',
        x='Model',
        order=modelD.keys(),
        hue='Model',
        hue_order=modelD.keys(),
        dodge=False,
        jitter=0.2,
        size=4,
        color='gray',
        edgecolor='black',
        linewidth=0.5,
        alpha=0.4,
        ax=ax)

        ax.set_title(cohort)
        ax.set_xlabel('')
        ax.set_ylabel('Out-of-sample '+yvar_model+' '+ r'$R^2$')

        ax.tick_params(axis='x', rotation=70)

    sns.despine()

    plt.tight_layout()
    #plt.savefig( f'{yvar_model}_Model_comparison.png',bbox_inches='tight',dpi=300)
    plt.show()
```

## -4 pairplot

```python
sns.set(style='ticks', context='notebook')

for cohort in cohortD.keys():
    tempL = []
    for model , pred in predictD.items():
        tempS = pred[cohort]['AgeGap'].copy()
        tempS.name = model
        tempL.append(tempS)
    tempDF = pd.concat(tempL, axis=1)
    tempDF = tempDF.select_dtypes(include='number')

    # PairGrid
    g = sns.PairGrid(tempDF, height=2)

    # 左下：Spearman相関の符号で色分け
    def scatter_corr(x, y, **kwargs):
        valid = np.isfinite(x) & np.isfinite(y)
        x = np.asarray(x)[valid]
        y = np.asarray(y)[valid]

        ax = plt.gca()

        if len(x) < 3 or np.std(x) == 0 or np.std(y) == 0:
            return

        rho, _ = spearmanr(x, y)

        color = 'red' if rho >= 0 else 'blue'
        ax.scatter(x, y, alpha=0.2, s=15, color=color)

    g.map_lower(scatter_corr)

    # 対角線：分布
    g.map_diag(sns.histplot, kde=True)

    # 右上：Spearman相関係数とp値
    def corrfunc(x, y, **kwargs):
        valid = np.isfinite(x) & np.isfinite(y)
        x = np.asarray(x)[valid]
        y = np.asarray(y)[valid]

        ax = plt.gca()

        if len(x) < 3 or np.std(x) == 0 or np.std(y) == 0:
            ax.text(
                0.5, 0.5, 'N/A',
                ha='center', va='center',
                transform=ax.transAxes)
            ax.set_axis_off()
            return

        rho, p = spearmanr(x, y)
        p_text = f'{p:.2e}' if p < 0.001 else f'{p:.3f}'

        ax.annotate(
            f'ρ = {rho:.3f}\np = {p_text}\nn = {len(x)}',
            xy=(0.5, 0.5),
            xycoords='axes fraction',
            ha='center',
            va='center',
            fontsize=15,
            color='red' if rho >= 0 else 'blue')
        ax.set_axis_off()

    g.map_upper(corrfunc)

    g.fig.suptitle(f'{cohort} - Spearman correlation', y=1.02, fontsize=20)
    g.fig.tight_layout()
    plt.show()
```

```python
sns.set(style='ticks', context='notebook')

for cohort in cohortD.keys():
    tempL = []
    for model , pred in predictS.items():
        tempS = pred[cohort]['AgeGap'].copy()
        tempS.name = model
        tempL.append(tempS)
    tempDF = pd.concat(tempL, axis=1)
    tempDF = tempDF.select_dtypes(include='number')

    # PairGrid
    g = sns.PairGrid(tempDF, height=2)

    # 左下：Spearman相関の符号で色分け
    def scatter_corr(x, y, **kwargs):
        valid = np.isfinite(x) & np.isfinite(y)
        x = np.asarray(x)[valid]
        y = np.asarray(y)[valid]

        ax = plt.gca()

        if len(x) < 3 or np.std(x) == 0 or np.std(y) == 0:
            return

        rho, _ = spearmanr(x, y)

        color = 'red' if rho >= 0 else 'blue'
        ax.scatter(x, y, alpha=0.2, s=15, color=color)

    g.map_lower(scatter_corr)

    # 対角線：分布
    g.map_diag(sns.histplot, kde=True)

    # 右上：Spearman相関係数とp値
    def corrfunc(x, y, **kwargs):
        valid = np.isfinite(x) & np.isfinite(y)
        x = np.asarray(x)[valid]
        y = np.asarray(y)[valid]

        ax = plt.gca()

        if len(x) < 3 or np.std(x) == 0 or np.std(y) == 0:
            ax.text(
                0.5, 0.5, 'N/A',
                ha='center', va='center',
                transform=ax.transAxes)
            ax.set_axis_off()
            return

        rho, p = spearmanr(x, y)
        p_text = f'{p:.2e}' if p < 0.001 else f'{p:.3f}'

        ax.annotate(
            f'ρ = {rho:.3f}\np = {p_text}\nn = {len(x)}',
            xy=(0.5, 0.5),
            xycoords='axes fraction',
            ha='center',
            va='center',
            fontsize=15,
            color='red' if rho >= 0 else 'blue')
        ax.set_axis_off()

    g.map_upper(corrfunc)

    g.fig.suptitle(f'{cohort} - Spearman correlation', y=1.02, fontsize=20)
    g.fig.tight_layout()
    plt.show()
```
