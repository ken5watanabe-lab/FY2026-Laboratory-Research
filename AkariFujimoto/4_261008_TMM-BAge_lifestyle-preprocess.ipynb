```python
tempDF1 = diDF.copy()

tempDF = pd.DataFrame({
    '発症歴あり人数': tempDF1.eq(1).sum(),
    '有効回答数': tempDF1.notna().sum(),})

tempDF['発症歴あり割合(%)'] = (tempDF['発症歴あり人数'] / tempDF['有効回答数'] * 100).round(2)

tempDF = tempDF.sort_values('発症歴あり人数', ascending=False)

display(tempDF)

# 病名ごとのサブプロット
tempI=tempDF1.eq(1).sum().sort_values(ascending=False).index
ncols = 6
nrows = math.ceil(len(tempI) / ncols)

fig, axes = plt.subplots(
    nrows, ncols,
    figsize=(ncols * 3, nrows * 2.5),
    sharey=True)

axes = axes.flatten()

for i, col in enumerate(tempI):
    ax = axes[i]

    # 0/1の人数を集計（欠損値は除外）
    counts = tempDF1[col].value_counts().reindex([0, 1], fill_value=0)

    ax.bar(['0', '1'], counts.values)
    ax.set_title(f'{col}\n(n={int(counts.sum())}, 1={int(counts[1])})',
                 fontsize=8)
    ax.tick_params(axis='both', labelsize=8)

    # 1の人数を棒の上に表示
    ax.text(1, counts[1], str(int(counts[1])),
            ha='center', va='bottom', fontsize=18)

# 余ったサブプロットを削除
for j in range(len(tempI), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()

```

```python
# ==========================================
# Tier 1 / Tier 2 の疾患群を作成
# ==========================================

tempDF1 = pd.DataFrame(index=diDF.index)

for tier in ['tier1', 'tier2']:

    for disease, cols in history_dict[tier].items():

        # diDFに存在する項目だけ使用
        cols_exist = []

        for keyword in cols:
            cols_exist.extend([col for col in diDF.columns if keyword in col])
        cols_exist = list(dict.fromkeys(cols_exist))

        if len(cols_exist) == 0:
            print(f'{disease}: 該当columnなし')
            continue

        # どれか1つでも「1」なら1
        tempDF1[disease] = diDF[cols_exist].eq(1).any(axis=1).astype(float)

        # 構成項目がすべてNaNなら、疾患群もNaN
        all_nan = diDF[cols_exist].isna().all(axis=1)
        tempDF1.loc[all_nan, disease] = np.nan


# ==========================================
# Tier情報を付与して集計
# ==========================================

tierD = {}

for tier in ['tier1', 'tier2']:
    for disease in history_dict[tier]:
        if disease in tempDF1.columns:
            tierD[disease] = tier


tempDF = pd.DataFrame({
    'Tier': pd.Series(tierD),
    '発症歴あり人数': tempDF1.eq(1).sum(),
    '有効回答数': tempDF1.notna().sum(),
})

tempDF['発症歴あり割合(%)'] = (
    tempDF['発症歴あり人数']
    / tempDF['有効回答数']
    * 100
).round(2)

tempDF = tempDF.sort_values(
    '発症歴あり人数',
    ascending=False
)

display(tempDF)


# ==========================================
# 疾患群ごとのサブプロット
# ==========================================

tempI = tempDF.index

ncols = 6
nrows = math.ceil(len(tempI) / ncols)

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(ncols * 3, nrows * 2.5),
    sharey=True
)

axes = np.atleast_1d(axes).flatten()

for i, col in enumerate(tempI):

    ax = axes[i]

    # 0/1の人数を集計（NaNは除外）
    counts = (
        tempDF1[col]
        .value_counts()
        .reindex([0, 1], fill_value=0)
    )

    ax.bar(['0', '1'], counts.values)

    ax.set_title(
        f'{col}\n'
        f'(n={int(counts.sum())}, 1={int(counts[1])})',
        fontsize=8
    )

    ax.tick_params(axis='both', labelsize=8)

    # 1の人数
    ax.text(
        1,
        counts[1],
        str(int(counts[1])),
        ha='center',
        va='bottom',
        fontsize=18
    )

# 余ったsubplotを削除
for j in range(len(tempI), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
