# Mapping OlinkID and Assay

```python
#Initial setup
import warnings
warnings.filterwarnings('ignore')
from IPython.display import display

import pandas as pd
import os
import numpy as np
```

```python
fileDir = "./Data/"
fileName = "pro_data.csv"

path = os.path.join(fileDir, fileName)
temp = pd.read_csv(path)
temp.set_index(temp.columns[0],inplace=True)
temp.index.name = None
pro_data = temp
display(pro_data)

fileDir = "./Data/"
fileName = "map_OID_Uni.csv"

path = os.path.join(fileDir, fileName)
IDdata = pd.read_csv(path)
IDdata.set_index(IDdata.columns[0],inplace=True)
IDdata.index.name = None
display(IDdata)

```

```python
temp = dict(
    zip(IDdata["OlinkID"], IDdata["UniProt"])
)
temp2 = pro_data.rename(
    columns=temp
)
display(temp2)
pro_data = temp2
```

```python
pro_data.to_csv("./Data/pro_data_clean_mapped.csv", index=True)
```
