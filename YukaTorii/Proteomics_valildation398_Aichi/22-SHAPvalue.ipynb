# SHAP value for each model

## CA19-9 + DEPlist

```python
dep = pro_data_dis[DEPlist.index].copy()
temp = ca199_dis[['logvalue']].copy()
temp = temp.rename(columns={'logvalue': 'CA19_9'})

X = pd.concat([temp, dep], axis=1)

print(type(X))
print(X.shape)
print(X.columns.tolist())
```

```python
model = modelsets_HER_DEP_ca199[0]
print(model.classes_)
explainer = shap.TreeExplainer(model)
shap_values = explainer(X)

print(type(shap_values))
print(shap_values.shape)
shap_values_array = shap_values.values[:, :, 1]
feature_names = np.array(X.columns)
shap.summary_plot(
    shap_values_array,
    X.values,
    feature_names=feature_names,
    max_display=20,
    show=False
)
plt.title("SHAP Summary Plot (HERPACC 400(Dis): DEPs + CA19-9)", fontsize=13, pad=15)
plt.tight_layout()
plt.show()
```
