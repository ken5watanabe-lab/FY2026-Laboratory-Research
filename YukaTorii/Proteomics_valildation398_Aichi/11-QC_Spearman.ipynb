# Import library

```python
import pandas as pd
from scipy.stats import spearmanr
import seaborn as sns
import matplotlib.pyplot as plt
import os
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial.distance import squareform
from sklearn.preprocessing import StandardScaler
```

# Import datasets

```python
fileDir = "./Data/"
fileName = "pro_data_common_dis.csv"

path = os.path.join(fileDir, fileName)
temp = pd.read_csv(path)
temp.set_index(temp.columns[0],inplace=True)
temp.index.name = None
display(temp)

#Z-score transformation
print('Before:')
display(temp.iloc[:, :15].describe())

temp2 = temp.copy()
scaler = StandardScaler(copy=True, with_mean=True, with_std=True)
tempA = scaler.fit_transform(temp2)#axis: column
tempDF = pd.DataFrame(data=tempA, index=temp2.index, columns=temp2.columns)

print('After:')
display(tempDF)
display(tempDF.iloc[:, :15].describe())

pro_data_dis = tempDF 
```

```python
fileDir = "./Data/"
fileName = "pro_data_common_val.csv"

path = os.path.join(fileDir, fileName)
temp = pd.read_csv(path)
temp.set_index(temp.columns[0],inplace=True)
temp.index.name = None
display(temp)

#Z-score transformation
print('Before:')
display(temp.iloc[:, :15].describe())

temp2 = temp.copy()
scaler = StandardScaler(copy=True, with_mean=True, with_std=True)
tempA = scaler.fit_transform(temp2)#axis: column
tempDF = pd.DataFrame(data=tempA, index=temp2.index, columns=temp2.columns)

print('After:')
display(tempDF)
display(tempDF.iloc[:, :15].describe())

pro_data_val = tempDF 
```

## メタデータ

### Validation

```python
fileDir = "./Data/"
fileName = "pro_sample.csv"

path = os.path.join(fileDir, fileName)
temp = pd.read_csv(path)
temp.set_index(temp.columns[0],inplace=True)
temp.index.name = None

temp["Cohort"] = "Validation"

display(temp)

pro_sample_val = temp
```

### Discovery

```python
fileDir = "../Proteomics/Data/"
fileName = "pro_sample.csv"

path = os.path.join(fileDir, fileName)
temp = pd.read_csv(path)
temp.set_index(temp.columns[0],inplace=True)
temp.index.name = None

temp["Cohort"] = "Discovery"

display(temp)

pro_sample_dis = temp
```

# DiscoveryとValidationを統合

## analyte data

```python
temp = pd.concat(
    [pro_data_val, pro_data_dis],
    axis=0,
    ignore_index=False  # インデックスを保持
)
display(temp)
pro_data = temp
```

## meta data

```python
display(pro_sample_dis)
display(pro_sample_val)
```

```python
# Discovery側
meta_dis = pro_sample_dis[[
    "fvage",
    "groupset2",
    "Cohort",
    "quant_drink_gr",
    "py",
    "hx_dm",
    "sex"
]].rename(columns={"groupset2": "CaCo_Status"})

# Validation側
meta_val = pro_sample_val[[
    "fvage",
    "CaCo_Status",
    "Cohort",
    "quant_drink_gr",
    "py",
    "hx_dm",
    "sex"
]]

# 縦に結合
pro_sample = pd.concat([meta_dis, meta_val], axis=0)

display(pro_sample)
```

```python
# Spearman correlation
corr = pro_data.T.corr(method="spearman")
print(corr)
```

```python

plt.figure(figsize=(12, 10))

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    square=True
)

plt.title("Spearman Correlation")
plt.show()
```

```python
sns.clustermap(
    corr,
    cmap="coolwarm",
    center=0,
    figsize=(12, 12)
)
```
