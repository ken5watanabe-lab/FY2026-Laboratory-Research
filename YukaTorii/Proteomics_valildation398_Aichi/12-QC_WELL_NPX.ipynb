```python
display(pro_data)
```

```python
corr = pro_data.T.corr(method="spearman")
```

```python
plate_colors = dict(
    zip(
        metadata["PlateID"].unique(),
        sns.color_palette("tab20", metadata["PlateID"].nunique())
    )
)

row_colors = pd.DataFrame({
    "Plate": metadata["PlateID"].map(plate_colors)
}, index=metadata.index)

row_palette = dict(
    zip(
        list("ABCDEFGH"),
        sns.color_palette("Set2", 8)
    )
)
col_palette = dict(
    zip(
        range(1, 13),
        sns.color_palette("viridis", 12)
    )
)
row_colors["Row(A~H)"] = metadata["WellRow"].map(row_palette)
row_colors["Column(1~12)"] = metadata["WellCol"].map(col_palette)
```

```python

sns.clustermap(
    corr,
    cmap="coolwarm",
    metric = 'euclidean',
    center=0,
    figsize=(14,14),
    method = "ward",
    row_colors=row_colors,
    col_colors=row_colors,
    xticklabels = False,
    yticklabels = False

)
```
