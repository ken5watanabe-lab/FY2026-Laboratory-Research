# Enrichment

## BP(Biological Process)

```r
head(module_tbl)
```

```r
modules <- unique(module_tbl$ModuleID)

ego_BP_list <- lapply(modules, function(mod) {
    
    proteins <- module_tbl %>%
        filter(ModuleID == mod) %>%
        pull(Symbol)
    
    enrichGO(
        gene = proteins,
        universe = module_tbl$Symbol,
        OrgDb = org.Hs.eg.db,
        ont = "BP",
        pAdjustMethod = "BH",
        pvalueCutoff = 1,
        qvalueCutoff = 1,
        minGSSize = 4,
        maxGSSize = 200,
        keyType = "SYMBOL",
        readable = TRUE
    )
})

names(ego_BP_list) <- modules
```

```r
head(as.data.frame(ego_BP_list$Turquoise))
```

```r
head(as.data.frame(ego_BP_list$Blue))
```

```r
head(as.data.frame(ego_BP_list$Yellow))
```

```r
ego_BP_all <- lapply(names(ego_BP_list), function(mod) {
    
    result <- as.data.frame(ego_BP_list[[mod]])
    
    if (nrow(result) > 0) {
        result$Module <- mod
        result <- result %>%
            dplyr::select(Module, everything())
    }
    
    result
}) %>%
    bind_rows()
```

```r
ego_BP_all
```

```r
ME_features <- c(
    "MEblue",
    "MEbrown",
    "MEgreen",
    # "MEgrey",
    "MEred",
    "MEturquoise",
    "MEyellow"
)

selected_modules <- str_to_title(
    str_remove(ME_features, "^ME")
)

ego_BP_selected <- ego_BP_all %>%
    filter(Module %in% selected_modules) %>%
    group_by(Module) %>%
    slice_head(n = 6) %>%
    ungroup()

View(ego_BP_selected)
```

```r
names(ego_MF_list)
```

```r
p_adjust_limit <- 0.05

p_adjust_scale <- scale_colour_gradient(
    low = "red",
    high = "blue",
    limits = c(0, p_adjust_limit),
    oob = scales::squish
)
for (me in ME_features) {
    
    module <- str_to_title(str_remove(me, "^ME"))
    
    p_BP <- dotplot(
        ego_BP_list[[module]],
        showCategory = 10,
        title = paste0(module, " - BP")
    ) +
        p_adjust_scale
    
    p_CC <- dotplot(
        ego_CC_list[[module]],
        showCategory = 10,
        title = paste0(module, " - CC")
    ) +
        p_adjust_scale
    
    p_MF <- dotplot(
        ego_MF_list[[module]],
        showCategory = 10,
        title = paste0(module, " - MF")
    ) +
        p_adjust_scale
    
    print(p_BP)
    print(p_CC)
    print(p_MF)
}
```

```r
library(enrichplot)
options(repr.plot.width = 12, repr.plot.height = 12)

# Cnetplotの実行

ekegg2 <- setReadable(ekegg, 
                           OrgDb = org.Hs.eg.db, 
                           keyType = "UNIPROT") # もし元のIDがUniProtなら "UNIPROT"

# showCategoryで表示するパスウェイ数を指定（上位5つなど）
p <- cnetplot(
    ekegg2, 
    showCategory = 6, 
    circular = TRUE,    # 円状に配置して見やすくする
    colorEdge = TRUE)   # パスウェイごとに線を色分け
p
```

```r
options(repr.plot.width=10, repr.plot.height=12)

# 1. データフレームに変換し、上位30個を抽出
temp <- as.data.frame(ego) %>%
  # 補正p値の小さい順にソートして上位30個を選択
  arrange(p.adjust) %>%
  slice_head(n = 30) %>%
  # 可視化のために -log10(p.adjust) を計算
  mutate(log_p = -log10(p.adjust),
        nGeneRatio = sapply(GeneRatio, function(x) {
            nums <- as.numeric(strsplit(x, "/")[[1]])
            return(nums[1]/nums[2])
        })
         )

# 2. ggplot2で棒グラフを作成
p <- ggplot(temp, aes(x = reorder(Description, nGeneRatio), y = nGeneRatio, fill = p.adjust)) +
  geom_col() +
  coord_flip() +  # 横向きの棒グラフにする
  scale_fill_viridis_c() +  
  labs(
    x = "GO Terms",
    y = "-log10(Adjusted P-value)",
    fill = "-log10(Adj P-value)",
    title = "Enriched GO Terms (Top 30)"
  ) +
  theme_minimal()

print(p)
```
